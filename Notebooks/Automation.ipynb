```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

orders = pd.read_csv("../Data/Cleaned data/orders.csv")
products = pd.read_csv("../Data/Cleaned data/products.csv")

# standardize column names
orders.columns = orders.columns.str.lower()
products.columns = products.columns.str.lower()

# merge datasets
df = orders.merge(products, on="product_id")

df.head()
```

```python
basket = pd.crosstab(df['order_id'], df['product_name'])
basket.head()
```

```python
# Convert Values to 0/1
basket = basket.apply(lambda x: x.apply(lambda y: 1 if y > 0 else 0))
basket.head()
```

```python
# Run Apriori Algorithm

frequent_items = apriori(basket, min_support=0.02, use_colnames=True)

frequent_items.head()
```

```python
# Generate Association Rules


rules = association_rules(frequent_items, metric="lift", min_threshold=1)

rules.sort_values(by="lift", ascending=False).head()
```

Business Insights:

Customers buying Cakes frequently purchase Candles together.

Gift combinations such as Flowers and Chocolates show strong association.

These product bundles can be used for cross-selling recommendations.

Retailers can create combo offers to increase average order value.
